```python
#fiilna()    (empty values fill)
data1=data.fillna(0)
```

```python
data1
```

```python
#empty values count in data1
data1.isnull().sum()
```

```python
data1.shape
```

```python
#drop()
data2=data1.drop(['CustomerID','Discount','ProductID','CustomerType','PaymentMethod','Return','CustomerType','SatisfactionScore','Quantity'],axis=1)
data2
```

```python
#rename()
data3=data2.rename(columns={'SalesChannel':'Sales Place','ProductCategory':'Product Name'})
data3
```

```python
#unique()
data3['Region'].unique()
```

```python
data3['Product Name'].unique()
```

```python
#nunique()
data3['Region'].nunique()
```

```python
data3['Product Name'].nunique()
```

```python
#groupby
data4=data3.groupby('Region')['Sales'].count()
data4
```

```python
#MATPLOT PLOTTING       (BAR CHART)
regions=data4.index.astype(str).tolist()
sales=data4.values.astype(float)
plt.figure()
plt.bar(regions,sales)
plt.title("REGION VS SALES")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
#bar chart    (product name vs sales)
plt.bar(data3["Product Name"],data3["Sales"])
plt.title("PRODUCT NAME VS SALES")
plt.xlabel("Product Name")
plt.ylabel("Sales")
plt.show()
```

```python
#bar chart    (sales place vs sales)
plt.bar(data3["Sales Place"],data3["Sales"])
plt.title("SALES PLACE VS SALES")
plt.xlabel("Sales Place")
plt.ylabel("Sales")
plt.show()
```

```python
#histogram   (sales)
plt.hist(data3['Sales'])
plt.title("SALES")
plt.show()
```
