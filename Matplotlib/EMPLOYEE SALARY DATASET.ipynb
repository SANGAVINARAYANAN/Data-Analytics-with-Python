```python
#MATPLOT                (SALARY Vs GENDER  ,   BAR CHART) 
plt.bar(data1['Gender'],data1['Salary'])
plt.title("SALARY Vs GENDER")
plt.xlabel("Gender")
plt.ylabel("Salary")
plt.show()
```

```python
          #SALARY VS AGE , BAR CHART
plt.bar(data1['Age'],data1['Salary'])
plt.title("SALARY VS AGE")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
  #(SALARY VS EXPERIENCE_YEARS,  SCATTER CHART) 
plt.scatter(data1['Experience_Years'],data1['Salary'])
plt.title("SALARY  VS  EXPERIENCE_YEARS")
plt.xlabel("Experience_Years")
plt.ylabel("Salary")
plt.show()
```

```python
#(EXPERIENCE_YEARS,    HISTOGRAM CHART)
plt.hist(data1['Experience_Years'])
plt.title("EXPERIENCE YEARS")
plt.show()
```
